```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
```

```python
ds = pd.read_csv('Crop_recommendation.csv')
ds.head()
ds.shape[0]
```

```python
ds.head()
```

```python
X = ds.loc[:,ds.columns.isin(['N','P','K','ph'])].values
```

```python
Y = pd.Categorical(ds['label']).codes
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test = train_test_split(X, test_size=0.4)
X_test.shape[0]
```

```python
Y_train, Y_test = train_test_split(Y, test_size=0.4)
```

```python
knn = KNeighborsClassifier(n_neighbors=21).fit(X_train,Y_train)
```

```python
prediction = knn.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(prediction,Y_test)
```

```python
from sklearn.metrics import classification_report
print(classification_report(prediction, Y_test))
```
